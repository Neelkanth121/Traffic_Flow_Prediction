### Exploratory Data Analysis :

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import numpy as np
```

```python
flow_df = pd.read_csv('../data/processed/flow_150_data.csv')
flow_df.head()
```

```python
plt.scatter(flow_df['avg_speed'],flow_df['flow'],s=1,color="orange")
plt.xlabel("avg speed")
plt.ylabel("Flow for 150s")
plt.grid()
plt.title("Avg Speed vs Flow for 150s")
```

```python
plt.scatter(flow_df['density'],flow_df['flow'],s=1,color="lightgreen")
plt.xlabel("density")
plt.ylabel("Flow for 150s")
plt.grid()
plt.title("Density vs Flow for 150s")
```

```python
plt.hist(flow_df['avg_speed'],edgecolor='black',facecolor="orange",bins=50);
plt.xlabel("avg speed")
plt.ylabel("Counts")
plt.grid(axis='y')
plt.title("Avg speed Distribution")
```

```python
plt.hist(flow_df['density'],edgecolor='black',facecolor="orange",bins=35);
plt.xlabel("Density")
plt.ylabel("Counts")
plt.grid(axis='y')
plt.title("Density Distribution")
```

```python
plt.hist(flow_df['flow'],edgecolor='black',facecolor="orange",bins=40);
plt.xlabel("Flow")
plt.ylabel("Counts")
plt.grid(axis='y')
plt.title("Flow Distribution")
```
